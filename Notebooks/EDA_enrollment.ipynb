```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent  # notebooks → project root
DATA_DIR = PROJECT_ROOT / "Data"
RAW_DATA_DIR = DATA_DIR / "Raw"
ENROLMENT_DATA_DIR = RAW_DATA_DIR / "enrolment"

print("Project root:", PROJECT_ROOT)
print("Enrolment dir exists:", ENROLMENT_DATA_DIR.exists())
```

```python
from statewise_eda import process_state_eda
```

```python
STATES_TO_ANALYZE = [
    'andaman_and_nicobar_islands', 'andhra_pradesh', 
    'arunachal_pradesh', 'assam', 'bihar', 'chandigarh', 
    'chhatisgarh', 'chhattisgarh', 'dadra_and_nagar_haveli_and_daman_an', 
    'dadra_and_nagar_haveli', 'daman_and_diu', 'delhi', 'goa', 
    'gujarat', 'haryana', 'himachal_pradesh', 'jammu_and_kashmir', 
    'jharkhand', 'karnataka', 'kerala', 'ladakh', 'lakshadweep', 
    'madhya_pradesh', 'maharashtra', 'manipur', 'meghalaya', 
    'mizoram', 'nagaland', 'odisha', 'puducherry', 'punjab', 
    'rajasthan', 'sikkim', 'tamil_nadu', 'telangana', 'tripura', 
    'uttar_pradesh', 'uttarakhand', 'west__bengal', 'west_bengal'
]
```

```python
# Directory for combined summaries
reports_dir = Path("Reports/Statewise_EDA")
reports_dir.mkdir(parents=True, exist_ok=True)

# Loop over states
all_summaries = ""
for state in STATES_TO_ANALYZE:
    print(f"Processing EDA for {state}...")
    summary = process_state_eda(state)
    all_summaries += summary + "\n" + "="*50 + "\n"
    print(f"Completed for {state}.\n")

# Save combined text file
combined_file = reports_dir / "all_states_eda_summary.txt"
with open(combined_file, 'w') as f:
    f.write(all_summaries)

print(f"All EDA summaries saved to {combined_file}")
```
